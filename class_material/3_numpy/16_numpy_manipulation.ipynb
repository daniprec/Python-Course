## Joining arrays

- `np.concatenate((a, b), axis=0)`: one on top of the other.
- `np.concatenate((a, b), axis=1)`: side by side.
- All dimensions except the joining one must match.

Compare with `np.sum(a, axis=0)`: concatenate **adds** along the axis, sum **squeezes** it.

## Splitting arrays

- `np.split(arr, 3)`: equal parts (error if not possible).
- `np.array_split(arr, 4)`: unequal parts allowed.
- `np.hsplit()` / `np.vsplit()`: by columns / by rows.
- A **list of positions** cuts where you want: `np.hsplit(stats, [2, 4])` gives attack, defence, discipline.

## Summary

- `reshape()` changes the arrangement, not the count. Reshape is **not** transpose.
- `flatten()` gives a 1D copy.
- `concatenate()`, `vstack()`, `hstack()`, `column_stack()` join arrays: the non-joining dimensions must match.
- `split()`, `array_split()`, `hsplit()`, `vsplit()` cut arrays, also at given positions.
