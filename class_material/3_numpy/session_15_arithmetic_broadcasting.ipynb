## 3. Broadcasting

`marks` is `(6, 11)`, `weights` is `(11,)`, and `marks * weights` works. Why?

- Compare shapes from the **right**.
- Compatible if equal, or one of them is 1.
- The smaller array is stretched.

```text
marks     (6, 11)
weights       (11,)  -> (1, 11) -> (6, 11)
```
