## 4. Operations with sets

| Mathematics | Python | Operator |
|---|---|---|
| $A \cup B$ | `A.union(B)` | <code>A &#124; B</code> |
| $A \cap B$ | `A.intersection(B)` | `A & B` |
| $A \setminus B$ | `A.difference(B)` | `A - B` |
| $A \,\triangle\, B$ | `A.symmetric_difference(B)` | `A ^ B` |
