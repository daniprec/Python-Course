## Changing the drawer

Sets are mutable. Use `add()` to add one item. Use `remove()` or `discard()` to remove an item.

`remove()` raises an error when the item is absent. `discard()` does nothing in that case.

## What can go inside a set?

Set members must be immutable. Numbers, strings, booleans, and immutable tuples work. Lists do not.

## Comparing groups

Given two sets `A` and `B`:

| Question | Operator |
|---|---|
| In either group | `A \| B` |
| In both groups | `A & B` |
| In A but not B | `A - B` |
| In exactly one group | `A ^ B` |
