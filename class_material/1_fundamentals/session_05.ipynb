## return inside loop


**Exercise.** Make the function `is_prime(m)` that returns True if a number `m` is prime, False otherwise.

```python
num = 7

def is_prime(num):
  """
  This function checks if a number is prime
  """
  if num < 2:
    return False

  # i = possible divisors
  for i in range(2, num):
    if num % i == 0:
      return False

  # if i reach this line
  # it is because I couldnt divide the number
  return True

print(is_prime(9))
```

# Calling Functions Inside Functions

**Exercise.** Implement a function `closest_lower_prime(m)` that takes a number `m` and returns the closest prime number that is lower than `m`.

```python
def closest_lower_prime(m):
  for i in range(m-1, 0, -1):
    if is_prime(i):
      return i


print(closest_lower_prime(22))  # 19
```

## Recursive Functions

**Exercise.** Create a factorial recursive function.

```python
def factorial(num):
  return num * factorial(num - 1)

print(factorial(5))
```
