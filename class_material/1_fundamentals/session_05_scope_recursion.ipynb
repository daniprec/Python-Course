### Local, enclosing, global, built-in

| Scope | Where |
|---|---|
| **L**ocal | inside the current function |
| **E**nclosing | inside a surrounding function |
| **G**lobal | at the top level of the notebook |
| **B**uilt-in | `print`, `len`, `range` |
