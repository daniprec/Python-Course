## Indexing and Slicing

```python
msg = "Hello world"

# square brackets are for indexing
first_char = msg[-1]

print(first_char)
```

```python
word = msg[6:11]

print(word)
```

```python
# start:stop
word = msg[:5]

print(word)
```

```python
word = msg[6:]

print(word)
```

```python
word = msg[:]

print(word)
```

```python
word = msg[6:]
print(word[2])
```

```python
print(msg[6:][2])
```

```python
print(msg[2][6:])
```

```python
x = msg[2]
x[6:]
```

```python
print(msg[0:12:3])
```

```python
nums = "123456789"

print(nums[1::2])
```

## Flow Control

if, else, elif

```python
grade = 10 # out of 10

print("You passed!")
```

```python
grade = 3

if grade > 5:
  print("You passed!")
```

```python
print(6 % 2)
```

```python
num = 5

rem = (num % 2)

if rem == 1:
  print("its odd!")

if rem == 0:
  print("its even!")
```

```python
num = 4

rem = (num % 2)

if rem == 1:
  print("its odd!")

  if rem == 0:
    print("its even!")
```

```python
num = 10

rem = (num % 2)

if rem == 1:
  print("its odd!")
else:
  print("its even!")
```

```python
# >= 5 pass
# < 5 fail
# 10 honors!

num = 6

if num >= 5:
  print("pass")

if num == 10:
  print("pass and honor!")

else:
  print("fail")
```
