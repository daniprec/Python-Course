## Continue and Break

[Back to the slides](https://docs.google.com/presentation/d/1fqrnB94OHVW3pIVe59A_3SMny-to2GCF/edit?usp=sharing&ouid=112793419067458804625&rtpof=true&sd=true)

```python
ls_students = ["Benito", "Andrea", "Clara", "Marta", "Arturo"]

for student in ls_students:
  if student == "Andrea":
    continue
  print(f"Hi {student}")
```

```python
ls_students = ["Benito", "Andrea", "Clara", "Marta", "Arturo"]

for student in ls_students:
  if student == "Andrea":
    break
  print(f"Hi {student}")
```
